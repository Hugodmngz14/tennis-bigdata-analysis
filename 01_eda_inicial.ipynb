```python
winners = matches[[
    "surface",
    "winner_ht",
    "w_ace",
    "w_df",
    "w_svpt",
    "w_1stIn"
]]

winners = winners.rename(columns={
    "winner_ht": "height",
    "w_ace": "aces",
    "w_df": "double_faults",
    "w_svpt": "service_points",
    "w_1stIn": "first_serve_in"
})
```

```python
winners["first_serve_pct"] = (
    winners["first_serve_in"] /
    winners["service_points"]
) * 100
```

```python
winners = winners.dropna()

winners = winners[
    (winners["height"] >= 160) &
    (winners["height"] <= 220)
]
```

```python
winners[[
    "height",
    "first_serve_pct",
    "aces",
    "double_faults"
]].corr()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    winners["height"],
    bins=20,
    kde=True
)

plt.title("Distribución de alturas")
plt.xlabel("Altura")
plt.ylabel("Frecuencia")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=winners,
    x="height",
    y="first_serve_pct",
    alpha=0.4
)

plt.title("Altura vs Primer Saque")
plt.xlabel("Altura")
plt.ylabel("Porcentaje Primer Saque")

plt.show()
```

# Conclusiones iniciales

- La altura presenta una correlación débil con el porcentaje de primer saque.
- Los jugadores más altos parecen generar más aces.
- La superficie influye significativamente en el rendimiento del saque.
